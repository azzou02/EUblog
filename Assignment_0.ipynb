```python
# Question 2 find top 20 best performer stocks - i.e. the stocks that provided the best return.
# if the stocks adj_close on the earliest date was X and the adj_close on the 2021-01-01 is y and the number of days elapsed between these dates is Z.
# Then return = Y/X/Z
# Sort the stocks by this return and find the top 20
```

```python
# find top 20 using first and last date of adj_close
data['date'] = pd.to_datetime(data['date'])

def find_stock_performance(symbol_data):
  symbol_data = symbol_data.sort_values('date')
  X = symbol_data['adj_close'].iloc[0]
  Y = symbol_data['adj_close'].iloc[-1]

  start_date = symbol_data['date'].iloc[0]
  end_date = symbol_data['date'].iloc[-1]
  Z = (end_date - start_date).days
  if Z <= 0 or X <= 0:
    return 0

  return Y/X/Z

results = data.groupby('symbol').apply(find_stock_performance, include_groups=False)
top_20 = results.sort_values(ascending=False).head(20)
print(top_20)
```

```python
# find top 20 using first date and 2021-1-1 of adj_close
data['date'] = pd.to_datetime(data['date'])
target = pd.Timestamp('2021-01-01')

def find_stock_performance(symbol_data):
    symbol_data = symbol_data.sort_values('date')
    X = symbol_data['adj_close'].iloc[0]

    # restrict to target time for Y
    upto = symbol_data[symbol_data['date'] <= target]
    if upto.empty:
        return pd.NA
    Y = upto['adj_close'].iloc[-1]

    start_date = symbol_data['date'].iloc[0]
    end_date = upto['date'].iloc[-1]
    Z = (end_date - start_date).days
    if Z <= 0 or X <= 0:
        return pd.NA

    return Y/X/Z

results = data.groupby('symbol').apply(find_stock_performance, include_groups=False).dropna()
top_20 = results.sort_values(ascending=False).head(20)
print(top_20)
```
